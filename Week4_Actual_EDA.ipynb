# Week 4: Actual EDA and Feature Engineering
This notebook analyzes `week4_eda_sample.csv`. Run it in the same folder as the CSV.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('week4_eda_sample.csv')
```

```python
df['event_time'] = pd.to_datetime(df['event_time'], utc=True, errors='coerce')
df.shape, df.head()
```

```python
df.isna().sum().to_frame('missing_count').assign(missing_percent=lambda x: x['missing_count']/len(df)*100)
```

```python
df['event_type'].value_counts()
```

```python
df['event_type'].value_counts().plot(kind='bar', title='Customer Event Counts'); plt.xlabel('Event type'); plt.ylabel('Records'); plt.tight_layout(); plt.show()
```

```python
df['price'].describe()
```

```python
df['price'].hist(bins=50); plt.title('Product Price Distribution'); plt.xlabel('Price'); plt.ylabel('Frequency'); plt.tight_layout(); plt.show()
```

```python
df['brand'].value_counts().head(10).sort_values().plot(kind='barh', title='Top 10 Brands by Event Count'); plt.xlabel('Records'); plt.tight_layout(); plt.show()
```

```python
df['event_hour'] = df['event_time'].dt.hour
df['event_hour'].value_counts().sort_index().plot(kind='line', marker='o', title='Customer Activity by Hour (UTC)'); plt.xlabel('Hour (UTC)'); plt.ylabel('Records'); plt.grid(alpha=.3); plt.tight_layout(); plt.show()
```

```python
df['event_type'].value_counts(normalize=True).mul(100).round(2)
```

```python
df.groupby('user_id').agg(event_count=('event_type','size'), unique_products=('product_id','nunique'), session_count=('user_session','nunique')).head()
```
